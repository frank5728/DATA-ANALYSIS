# SECTION A: CREATING PANDAS DATAFRAMES

A1. Creating a DataFrame from a List

```python
import pandas as pd
```

```python
# Create a list containing student records
data = [
    ['241001','ADILI', 14],
    ['241002','JOSEPH', 15],
    ['241003','CAROL', 13],
    ['241004','FRANK', 16],
    ['241005','ATHANAS', 18]
]
```

```python
# Convert the list into a DataFrame and Assigning appropriate column names
df = pd.DataFrame(data, columns=['RegNo:','Name', 'Age'])
```

```python
# Display the DataFrame
print("Student Records DataFrame:")
print(df)
```

This output shows that we successfully converted a Python list into a Pandas DataFrame. Each inner list represents one student, while the column names are Registration Number, Name, and Age.

A2. Creating a DataFrame from a Dictionary

```python
# Create a dictionary containing student information
data = {
    "RegNo": [241001, 241002, 241003, 241004, 241005],
    "Name": ["ADILI", "JOSEPH", "CAROL", "FRANK", "ATHANAS"],
    "Age": [14, 15, 13, 16, 18]
}
```

```python
# Convert the dictionary into a DataFrame
record = pd.DataFrame(data)
```

```python
# Display the DataFrame
print("Student Records Dictionary:")
display(record)
```

Here, we created a DataFrame from a dictionary. The dictionary keys became the column names, while the values became the data in each column

```python
# Display the data types of each column in the DataFrame
record.dtypes
```

This output shows the data type of each column. Registration number and age are integers, while the student's name is stored as a string.

A3. Creating a DataFrame from Series

```python
# Create three Pandas Series
student_names = pd.Series(["ADILI", "JOSEPH", "CAROL", "FRANK", "ATHANAS"])
student_ages = pd.Series([14, 15, 13, 16, 18])
student_scores = pd.Series([75, 82, 68, 91, 79])
```

```python
# Combine the Series into a DataFrame
students_from_series = pd.DataFrame({
    "Student Name": student_names,
    "Age": student_ages,
    "Score": student_scores
})
```

```python
# Display the DataFrame
print("DataFrame created from Series:")
display(students_from_series)
```

This output shows that we created three Pandas Series for names, ages, and scores and then combined them into one DataFrame. The result is a structured table containing information about each student.

# 3. Data Cleaning

A. Check Missing Values

Training dataset

```python
print("Missing values in training dataset:")
```

```python
train.isnull().sum()
```

Testing dataset

```python
print("Missing values in testing dataset:")
```

```python
test.isnull().sum()
```

Here we are checking the dataset for missing information. Missing values are important because they can affect our analysis and may cause errors when performing calculations

 B. Check Duplicates

```python
print("Duplicate rows in training:", train.duplicated().sum())
```

```python
print("Duplicate rows in testing:", test.duplicated().sum())
```

We also checked for duplicate records. The result showed that there were no duplicate rows, so no duplicate records needed to be removed.

C. Handle Missing Ages

```python
train["Age"] = train["Age"].fillna(train["Age"].median())
```

Checking data consistency 

```python
train.isnull().sum()
```

This output confirms that the data-cleaning process was successful because every column now contains zero missing values.

```python
test["Age"] = test["Age"].fillna(train["Age"].median())
```

Checking data consistency

```python
test.isnull().sum()
```

This output confirms that the data-cleaning process was successful because every column now contains zero missing values.

D. Handle Missing Categorical Values

```python
train["Embarked"] = train["Embarked"].fillna(train["Embarked"].mode()[0])
```

```python
train["Cabin"] = train["Cabin"].fillna("Unknown")
```

Checking Data Consistency

```python
train.isnull().sum()
```

```python
test["Cabin"] = test["Cabin"].fillna("Unknown")
```

```python
test["Fare"] = test["Fare"].fillna(train["Fare"].median())
```

Checking Data Consistency

```python
test.isnull().sum()
```

After cleaning the data, the missing values were handled using appropriate methods. For example, missing numerical values such as age were replaced using the median, while missing categorical information was handled appropriately.

E. Convert Appropriate Variables to Categorical Data Types

```python
categorical_columns = ["Pclass", "Sex", "Embarked"]
```

```python
for column in categorical_columns:
    train[column] = train[column].astype("category")
    test[column] = test[column].astype("category")
```

```python
train["Survived"] = train["Survived"].astype("category")
```

We converted variables that represent categories into categorical data types. This makes the data more appropriate for grouping and categorical analysis.

# 5. Grouping

A. Survival Rate by Gender

```python
train["Survived"] = train["Survived"].astype(int)
```

```python
survival_by_gender = train.groupby("Sex")["Survived"].mean()
```

```python
print("Survival Rate by Gender:")
print(survival_by_gender)
```

This output shows the survival rate according to gender. About 74.20 percent of female passengers survived, compared with only 18.89 percent of male passengers. Therefore, females had a much higher survival rate than males in this dataset.

B. Survival Rate by Passenger Class

```python
survival_by_class = train.groupby("Pclass")["Survived"].mean()
```

```python
print("Survival Rate by Passenger Class:")
print(survival_by_class)
```

This output shows the survival rate according to passenger class. First-Class passengers had the highest survival rate at 62.96 percent, while Third-Class passengers had the lowest at 24.24 percent.

C. Average Fare by Passenger Class

```python
average_fare_class = train.groupby("Pclass", observed=True)["Fare"].mean()
```

```python
print(average_fare_class)
```

This output shows that First-Class passengers paid the highest average fare, approximately 84.15. Second-Class passengers paid about 20.66, while Third-Class passengers paid about 13.68.

D. Average Age by Gender

```python
average_age_gender = train.groupby("Sex", observed=True)["Age"].mean()
```

```python
print(average_age_gender)
```

This output compares the average age of male and female passengers. Female passengers had an average age of approximately 27.93 years, while male passengers had an average age of approximately 30.14 years.

# 6. Combining Training and Testing Datasets

```python
combined = pd.concat(
    [train, test],
    ignore_index=True
)
```

```python
print("\n===== COMBINED DATASET =====")
print(combined.head())
print("Combined shape:", combined.shape)
print("Combined columns:")
print(combined.columns.tolist())
```

After combining the two datasets, we obtained 1,309 passenger records. The Survived column is available for the training records, while it is not available for the testing records.

Why is pd.concat() Appropriate?

Finally, we combined the training and testing datasets using pd.concat(). This is appropriate because both datasets contain information about Titanic passengers and have a similar structure. The rows were combined vertically, and the index was reset using ignore_index=True.
