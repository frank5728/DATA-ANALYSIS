                                         DATA TRANSFORMATION

```python
#Study-time categories
df["study_time_category"] = pd.cut(
    df["study_time_hours"],
    bins=[0, 2, 4, 6, np.inf],
    labels=["0-2 hours", "2-4 hours", "4-6 hours", "6+ hours"],
    right=False
)

print(df[["study_time_hours", "study_time_category"]].head())
```

The result show diferent time category for example 4.0 study time hours range in category of 4-6 hours and 6.3 study time hours range in 6+ hours 

```python
#Attendance categories
df["attendance_category"] = pd.cut(
    df["attendance_percent"],
    bins=[0, 70, 80, 90, 101],
    labels=["Below 70%", "70-80%", "80-90%", "90%+"],
    right=False
)

print(df[["attendance_percent", "attendance_category"]].head())
```

The result show the attendace percentage based in attendes category where by 85.3 belong to 80-90% attendace category

```python
#Performance status
df["performance_status"] = np.where(
    df["final_grade"].isin(["A", "B", "C"]),
    "Pass",
    "At Risk"
)

print(df["performance_status"].value_counts())
```

The result show the number of student that can pass is 899 and the number of students that are in risk to fail is 101

```python
#Score improvement
df["score_change"] = (
    df["final_exam_score"] - df["previous_grade"]
)
print(df[["previous_grade", "final_exam_score", "score_change"]].head())
```

The result show the change score by comparing the score of previous and score of final exam score

                  KPI ANALYSIS

```python
# Total number of students
total_students = len(df)

# Average final exam score
average_final_score = df["final_exam_score"].mean()

# Average previous grade
average_previous_grade = df["previous_grade"].mean()

# Average study time
average_study_time = df["study_time_hours"].mean()

# Average attendance
average_attendance = df["attendance_percent"].mean()

# Average sleep hours
average_sleep = df["sleep_hours"].mean()

# Pass rate
# A, B and C are considered Pass
pass_rate = (
    df["final_grade"].isin(["A", "B", "C"]).mean()
) * 100

# At-risk rate
# D and F are considered At Risk
at_risk_rate = (
    df["final_grade"].isin(["D", "F"]).mean()
) * 100

# Internet access rate
internet_access_rate = (
    df["internet_access"] == "Yes"
).mean() * 100

# Score improvement
average_score_change = (
    df["final_exam_score"] - df["previous_grade"]
).mean()
```

                   DISPLAY KPIs

```python
print("========== BUSINESS KPIs ==========")

print("Total Students:",
      total_students)

print("Average Final Score:",
      round(average_final_score, 2))

print("Average Previous Grade:",
      round(average_previous_grade, 2))

print("Average Study Time:",
      round(average_study_time, 2),
      "hours")

print("Average Attendance:",
      round(average_attendance, 2),
      "%")

print("Average Sleep:",
      round(average_sleep, 2),
        "hours")

print("Pass Rate:",
      round(pass_rate, 2),
      "%")

print("At-Risk Rate:",
      round(at_risk_rate, 2),
      "%")

print("Internet Access Rate:",
      round(internet_access_rate, 2),
      "%")

print("Average Score Change:",
      round(average_score_change, 2))
```

                DATA VISUALIZATION

                     Grade distribution

```python
#Grade distribution
grade_order = ["A", "B", "C", "D", "F"]

grade_counts = (
    df["final_grade"]
    .value_counts()
    .reindex(grade_order, fill_value=0)
)

plt.figure(figsize=(8, 5))

grade_counts.plot(kind="bar")

plt.title("Students by Final Grade")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
#Study time vs final score
study_scores = df.groupby(
    "study_time_category",
    observed=False
)["final_exam_score"].mean()

plt.figure(figsize=(8, 5))

study_scores.plot(kind="bar")

plt.title("Average Final Score by Study Time")
plt.xlabel("Study Time")
plt.ylabel("Average Final Exam Score")

plt.tight_layout()
plt.show()
```

```python
#Attendance vs final score
plt.figure(figsize=(8, 5))

plt.scatter(
    df["attendance_percent"],
    df["final_exam_score"]
)

plt.title("Attendance vs Final Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Exam Score")

plt.tight_layout()
plt.show()
```

```python
# Internet access vs performance
internet_scores = df.groupby(
    "internet_access"
)["final_exam_score"].mean()

plt.figure(figsize=(7, 5))

internet_scores.plot(kind="bar")

plt.title("Internet Access vs Final Exam Score")
plt.xlabel("Internet Access")
plt.ylabel("Average Final Exam Score")

plt.tight_layout()
plt.show()
```

```python
# Part-time job vs performance
job_scores = df.groupby(
    "part_time_job"
)["final_exam_score"].mean()

plt.figure(figsize=(7, 5))

job_scores.plot(kind="bar")

plt.title("Part-Time Job vs Final Exam Score")
plt.xlabel("Part-Time Job")
plt.ylabel("Average Final Exam Score")

plt.tight_layout()
plt.show()
```
